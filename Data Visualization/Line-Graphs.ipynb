## Line Graphs

A line plot answers one question, "How does something change across an ordered variable,
especially time in this instance e.g where where are analyzing sales over a specified period
of time.

Line plots are particularly powerful for;

- sales over time
- stock levels over time
- temperature
- patient measurements over time
- production output
- website traffick
- financial prices
- medication consumption

It represents observations connected according to an variable. It is especially useful for
displaying trends over time. It emphasizes continutiy of that variable over a specified period
of time. Time series data is oen of the most common applications of line graphs


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

daily_transcations = df["SaleDate"].value_counts().sort_index()

first_ten = daily_transcations.head(10)

fig, ax = plt.subplots(
    figsize = (9,6)
)

ax.plot(
    first_ten.index,
    first_ten.values, 
    marker = "o"

)

ax.set_title("Transcations trend over a 10-day period")
ax.set_xlabel("Dates")
ax.set_ylabel("Number of transcations")

ax.grid(
    axis = "y"
)

ax.tick_params(
    axis = "x",
    rotation = 60
)

fig.tight_layout()
```

### Markers in Line plots

A line connects observations to emphasize continuity. Markers can be added to
identify the individual observations represented by the line

In Matplotlib, markers can be specified using the "marker" parameter

for examples
marker = "o"

adds circular makers at each observation


#### Let's take the same dataset and ask another question;

"How many transcations occured in each month?"


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

df["Month"] = df["SaleDate"].dt.to_period("M")

monthly_transcations = (
    df["Month"].value_counts().sort_index()
)

fig, ax = plt.subplots(
    figsize = (8,6)
)

ax.plot(
    monthly_transcations.index.astype(str),
    monthly_transcations.values,
    marker = "o"
)

ax.set_title("Monthly sales in the FY-2025")
ax.set_xlabel("Month")
ax.set_ylabel("Number of Sales")

ax.grid(
    axis = "y"
)

ax.tick_params(
    axis = "x",
    rotation = 60
)
```
