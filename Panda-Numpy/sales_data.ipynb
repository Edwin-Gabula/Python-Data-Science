```python
import pandas as pd
import numpy as np

sales = pd.read_excel(
    "pharmacy_sales.xlsx"
)


#selecting specific columns within a dataset

sales["Medicine"]

type(sales["Medicine"])

type(sales[["Medicine", "Quantity", "UnitPrice", "Insurance"]])

sales[["Medicine", "Quantity", "UnitPrice", "Insurance"]]

type(sales[["Medicine"]])


#indexing using location (loc)

sales.loc[3]

sales.loc[[1,3,4,6,7,14]]

sales.loc[2:8]

sales.loc[2:8, ["CustomerName", "Medicine", "Quantity", "Insurance"]]

sales.loc[4, "Medicine"]


#using the iloc method to select particular portions of the dataset

sales.loc[0:14, ["InvoiceID", "CustomerID", "Gender", "Age", "Medicine", "Quantity"]]

sales.iloc[[0,4,5,6,7,9,12]]

sales.iloc[0:14]

sales.iloc[2:12, 1:6]

sales["UnitPrice"] > 5000


sales[
    sales["UnitPrice"] > 5000
    ]

     #include customers with age over 40 years
sales [
    sales["Age"] > 40
]


#getting the number of individuals meeting that criteria
len(sales [
    sales["Age"] > 40
])

sales[
    sales["Branch"] == "Kampala"
]

#combining conditions

sales [
    (sales["Age"] > 40) & (sales["Branch"] == "Kampala") & (sales["PaymentMethod"] == "Insurance")
]


#the | means or
sales [
    (sales["Branch"] == "Kampala") | (sales["Branch"] == "Entebbe")
]

#setting the not include operators...
sales [
    ~ (sales["Branch"] == "Kampala")
]

#listings for antibiotics only

sales [
    sales["Category"] == "Antibiotic"
]

# meddicines with low stock

sales [
    sales["StockAvailable"] < 200
]


#checking for females

sales [
    sales["Gender"] == "Female"
]

#antibiotics with low stock

sales [
    (sales["Category"] == "Antibiotic") & (sales["StockAvailable"] < 200)
]


#samples questions

sales [
    sales["Age"] >= 30
]

sales [
    sales["Category"] == "Analgesic"
]

#customers from Wakiso who paid by cash

sales [
    (sales["Branch"] == "Wakiso") & (sales["PaymentMethod"] == "Cash")
]

sales [
    sales["Discount"] > 0
]

sales [
    ~ (sales["Branch"] == "Mukono")
]

# sorting datasets by column values

sales.sort_values(
    by = "UnitPrice", ascending=False
)

sales.sort_values(
    by = ["Branch", "Age"], ascending=[True, False]
)

sales.sort_values(
    by = "UnitPrice", ascending=False
)

expensive = sales[
    sales["UnitPrice"] > 5000
]

expensive.sort_values(
    by = "UnitPrice", ascending = False
)

insurance_paid = sales[
    (sales["PaymentMethod"] == "Insurance") & (sales["Category"] == "Antibiotic") & (sales["UnitPrice"] > 5000)
]

insurance_paid.sort_values(
    by = "UnitPrice", ascending=False

)


#trying to plot some basic graphs

import matplotlib.pyplot as plt

sales.plot(
    x = "Medicine",
    y = "UnitPrice",
    #kind="bar",
    figsize=(10,5),
    kind = "bar",
    title="Unit Price of Medicines"
)

plt.xlabel("Medicines"
           )
plt.ylabel("Unit Price (UGX)")
plt.xticks(rotation=90)
plt.show()


```
