## Data Visualization with Pandas and Matplotlib

- Data visualization is the process of representing data graphically so thatpatterns, relationships, distribution,
  trends and unsual observations can be identified easily

- In this chapter, we will combine the pandas skills learned with matplotlib to explore our
  pharmacy dataset visually. The important principle is;

  Data ---> Pandas ---> Transformation ---> Visualization ---> Interpretation

- Pandas is primarily responsible for loading, cleaning, selecting, filtering, grouping and transforming our data.
  Matplotlib is responsbile for creating and controlling the graphical representation.

- Pandas also provides convenient plotting methods that internally use matplotlib. Therefore, visualization is not replacement for Pandas. Instead, visualization builds upon the data manipulation skills.


```python
#importing the various tools needed in our data visualization and setting the data

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


#loading our data 
"""
before creating a visualization, the data must be loaded into a dataframe. we have used pd.Pharmacy_Sales_Dataset_2000_Records.xlsx
to read our excel workbook. the resulting object that we manipulate and visualize.

We should also inspect the dataset before we start plotting, and we achieve by using Pandas library
"""

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

# asking, how many records exist for each medicine

df["Medicine"].value_counts()


# trying out to plot medicines by a bar graph
"""
pandas does provide a convenient '.plot()' method for creating visualization directly from series
and DataFrame obejcts
"""

df["Medicine"].value_counts().plot(kind = "bar")  #the height of each bar represents the frequency of each of the medicines within the dataset

plt.title("Number of Records by Medicine")
plt.xlabel("Medicine")
plt.ylabel("Number of Records")
```

## Creating a Bar Plot

Matplotlib's 'plt.bar()' function creates a bar chart from two primary components

- plt.bar(x, height)
  The first argument represents he categories plotted along the x-axis
  The second argument represents the numerical values determining the height of the bars

In our examples above;
X = medicine names
height = medicine_count.values()


```python
# now doing an exercise to see if I have mastered my skills

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# importing the excel file

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

df.tail(10)

medicine_counts = df["Medicine"].value_counts()
branch_counts = df["Branch"].value_counts()

medicine_counts
branch_counts

# plotting two graphs side by side

fig, axes = plt.subplots(
    1,2,
    figsize = (10,8),
    layout = "constrained"
)

axes[0].bar(
    medicine_counts.index,
    medicine_counts.values
)

axes[1].bar(
    branch_counts.index,
    branch_counts.values
)

axes[0].set_title(
    "Sales by Medicine"
)

axes[1].set_title(
    "Sales by Branch"
)

axes[0].set_xlabel(
    "Medicine"
)

axes[0].set_ylabel(
    "Number of transcations"
)

axes[1].set_xlabel(
    "Branches"
)

axes[1].set_ylabel(
    "Number of transcations"
)

fig.tight_layout()
```

### Anatomy of a matplotlib plot

A matplotlib visualization consists of several graphical components.

The figure represents the overall canvas, while the axes represents the individual plotting \
region. An axes can contain

- plotted data
- a title
- an x-axis
- a y-axis
- axis labels
- tick marks
- tick labels
- grid lines
- spines
- legends and annotations

Understanding these components allows us to control a visualization systematically rather than
treating the graph as a single object


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

medicine_counts = df["Medicine"].value_counts()

fig, ax = plt.subplots(
    figsize = (8,6)
)

ax.bar(
    medicine_counts.index,
    medicine_counts.values
)

ax.set_xlabel("Medicines")
ax.set_ylabel("Number of Transcations")
ax.tick_params(
    axis = "x",
    rotation = 60
)

plt.tight_layout()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Pharmacy_Sales_Dataset_2000_Records.xlsx"
)

branch_counts = df["Branch"].value_counts()

fig, ax = plt.subplots(
    figsize = (8,6)
)

ax.bar(
    branch_counts.index,
    branch_counts.values
)

ax.set_xlabel("Branch")
ax.set_ylabel("Number of Transcations")
ax.set_title("Number of transcations per Medicine")

ax.tick_params(
    axis = "x",
    rotation = 60
)


# enabling grid lines... to be able to read values from the y-axis
ax.grid(
    axis = "y",
)

fig.tight_layout()
```
