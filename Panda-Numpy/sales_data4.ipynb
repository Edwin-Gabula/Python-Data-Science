```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_excel(
    "pharmacy_sales.xlsx"
)

female = data[
    data["Gender"] == "Female"
]

female["Age"].plot(
    kind = "hist",
    bins = 5,
    title="Age Distribution among females"
)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()






```
